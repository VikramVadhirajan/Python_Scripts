```python
import pandas as pd
csv_file = r"C:\Users\vikram.vadhirajan\OneDrive - Trico\Documents - Product Support India Romania\Catalog\EPR\Sales\Product to Customer BPI ASC IBI.csv"
```

```python
# Read chunk size
chunk_size = 200_000
```

```python
# scan_csv is lazy and memory-efficient
df = pd.read_csv(csv_file)
```

```python
df
```

```python
# Create a list of DataFrames by splitting the original DataFrame
df_chunks = [df.iloc[i:i + chunk_size] for i in range(0, len(df), chunk_size)]
```

```python
with pd.ExcelWriter('SPlitExcel.xlsx') as writer:  # doctest: +SKIP
    for i, chunk in enumerate(df_chunks):
        sheet_name = f"Chunk_{i+1}"  # Naming each sheet dynamically
        chunk.to_excel(writer, sheet_name=sheet_name, index=False)
```
