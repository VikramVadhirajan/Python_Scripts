```python
import xml.etree.ElementTree as ET
tree = ET.parse(r"20250113_025803.xml")
root = tree.getroot()
```

```python
import pandas as pd
```

```python
# Create a list to store dictionaries representing each item
Item_List = []
# Iterate through each <item> element
for item_elem in root.findall('<Items>'):
    item_dict = {}
    item_dict['id'] = item_elem.get('id')
    for child_elem in item_elem:
        item_dict[child_elem.tag] = child_elem.text
    Item_List.append(item_dict)
```

```python
df = pd.DataFrame(Item_List)
```
