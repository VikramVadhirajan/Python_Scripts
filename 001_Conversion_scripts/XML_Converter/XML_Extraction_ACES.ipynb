```python
import xml.etree.ElementTree as ET
import pandas as pd

# Load the XML file (update the filename if needed)
xml_file = r"ACES_2.xml"
```

```python
import xml.etree.ElementTree as ET

tags = set()

for event, elem in ET.iterparse(xml_file, events=("start",)):
    tags.add(elem.tag)

print(sorted(tags))

FullColumnList=list(sorted(tags))
```

```python
rows = []

for event, elem in ET.iterparse(xml_file, events=("end",)):

    if elem.tag == "App":

        row = {}

        # App attributes
        for k, v in elem.attrib.items():
            row[f"App_{k}"] = v

        # Child elements
        for child in elem:

            # attributes
            if child.attrib:
                for attr, val in child.attrib.items():
                    row[f"{child.tag}_{attr}"] = val

            # text values
            if child.text and child.text.strip():
                row[child.tag] = child.text.strip()

            # handle nested elements (Qual, param, etc.)
            for sub in child:
                if sub.text and sub.text.strip():
                    row[f"{child.tag}_{sub.tag}"] = sub.text.strip()

        rows.append(row)

        elem.clear()

df = pd.DataFrame(rows)
```

```python
ConvertedColumnList=df.columns.tolist()
```

```python
MissedCoulumns = list(set(FullColumnList) - set(ConvertedColumnList))
MissedCoulumns
```
