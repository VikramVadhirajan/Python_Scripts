```python
def split_with_second_delimiter(text, delimiter):
  """
  Splits a text only once with the second occurrence of the given delimiter.

  Args:
      text: The text to be split.
      delimiter: The delimiter to split on.

  Returns:
      tuple: A tuple containing the parts before and after the second delimiter,
              or (text, '') if no second delimiter is found.
  """

  # Find the first and second occurrences of the delimiter
  first_index = text.find(delimiter)
  second_index = text.find(delimiter, first_index + 2)

  # Check if at least two delimiters exist
  if second_index != -1:
    # Split the text based on the second delimiter
    return text[:second_index], text[second_index + 1:]
  else:
    # No second delimiter found, return the original text
    return text, ''

# Example usage
text = "2013 GMC Sierra 1500 WT 5.3L, V8, Flex, VIN 0, Des LMG, 16V, USA/Canada"
delimiter = ','

result = split_with_second_delimiter(text, delimiter)

print(result[0])  # Output: ('apple, banana', 'orange,grape')
```
