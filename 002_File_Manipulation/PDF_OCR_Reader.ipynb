```python
# Install required libraries
# !pip install PyPDF2 pandas openpyxl pdfplumber

# Import necessary libraries
import pdfplumber
import pandas as pd
```

```python
# Function to extract text from PDF
def extract_text_from_pdf(pdf_path):
    with pdfplumber.open(pdf_path) as pdf:
        text = ''
        for page in pdf.pages:
            text += page.extract_text()
    return text
```

```python

# Function to parse the extracted text into structured data
def parse_price_list(text):
    # Split text into lines
    lines = text.split('\n')
    
    # Assuming the PDF has two columns: Product and Price, split lines accordingly
    products = []
    prices = []
    
    for line in lines:
        if 'Product' in line or 'Price' in line:  # Skip header lines if needed
            continue
        # Assuming the product and price are separated by a space or tab
        parts = line.split()  # Adjust this split logic based on your PDF's format
        if len(parts) >= 2:
            product = ' '.join(parts[:-1])
            price = parts[-1]
            products.append(product)
            prices.append(price)
    
    # Create a DataFrame
    df = pd.DataFrame({
        'Product': products,
        'Price': prices
    })
    
    return df
```

```python

# Main function to process the PDF and export to Excel
def pdf_to_excel(pdf_path, excel_path):
    # Step 1: Extract text from PDF
    text = extract_text_from_pdf(pdf_path)
    
    # Step 2: Parse text to extract price list
    df = parse_price_list(text)
    
    # Step 3: Export DataFrame to Excel
    df.to_excel(excel_path, index=False)
    print(f"Price list has been saved to {excel_path}")
```

```python

# Example usage
pdf_path = r"C:\Users\vikram.vadhirajan\OneDrive - Trico\FBG\99_Downloads\Pricelist 2024.pdf"  # Replace with your actual PDF file path
excel_path = 'price_list.xlsx'  # Desired output Excel file path

pdf_to_excel(pdf_path, excel_path)
```

```python
import pdf2image
from PIL import Image
import pytesseract

def extract_text_from_pdf(pdf_path):
    """Extracts text from a PDF using OCR."""

    images = pdf2image.convert_from_path(pdf_path)

    text = ""
    for image in images:
        text += pytesseract.image_to_string(image)

    return text

if __name__ == "__main__":
    pdf_file = pdf_path
    extracted_text = extract_text_from_pdf(pdf_file)
    print(extracted_text)
```
