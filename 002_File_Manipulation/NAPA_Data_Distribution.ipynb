```python
import pandas as pd
import numpy as np
from tqdm import tqdm
from IPython.display import clear_output
```

# Mandatory Data

```python
Maodatory_Filling=r"path/filename.xlsx"
Madatory_output_location=r"folder"
df_MF=pd.read_excel(Maodatory_Filling,keep_default_na=False, na_values=[''])
df_MF
```

```python
df_MF.columns
```

```python
df_MF=df_MF[df_MF["Process_Step"]=="Smartsheet"]
df_MF=df_MF[['Brand','Product Type', 'SKU', 'Attribute','Value_Combined', 'Filled Date']]
```

```python
Brands=df_MF["Brand"].unique().tolist()
ProductTypes=df_MF["Product Type"].unique().tolist()
```

```python
for Brand in tqdm(Brands):
    with pd.ExcelWriter(rf"{Madatory_output_location}\\{Brand}_Mandatory_Data.xlsx") as writer:
        for ProductType in tqdm(ProductTypes):
            print(Brand, ProductType)
            df_temp=df_MF[(df_MF["Brand"]==Brand) & (df_MF["Product Type"]==ProductType)]
            pivot_df=df_temp.pivot(index=['Product Type','SKU'], columns='Attribute', values='Value_Combined').reset_index()
            if pivot_df.empty:
                continue
            ProductType=ProductType.replace('/', '_').replace(' ', "")
            if len(ProductType) > 31:
                ProductType = ProductType[:31]
            pivot_df.to_excel(writer, sheet_name=str(ProductType), index=True)
    clear_output()
```
