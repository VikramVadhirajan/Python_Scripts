```python
import os
import pandas as pd
from pathlib import Path
import warnings
warnings.filterwarnings('ignore')
from IPython.display import clear_output
```

```python
default_jupytext_formats = "ipynb,py:percent"
```

```python
folder=input("paste your reading folder:")
search_text='WarrantyByItem'
```

```python
file_list=[]
for (root, dirs, file) in os.walk(folder):
    for f in file:
        if ('.xlsx') in f:
            if search_text in f:
                file_list.append(f)
file_list
```

```python
file_link=[]

for i in range(len(file_list)):
     for r,d,f in os.walk(folder):
          for files in f:
               if files == file_list[i]:
                    file_link.append(os.path.join(r,files))
```

```python
file_link
```

```python
df=pd.DataFrame()
for file in file_link:
    df1=pd.read_excel(file)
    Brand=file.split("-")[1]
    df1["SourceName"]=file.split("-")[1]
    df=pd.concat([df,df1])
    print(f"{Brand} is appended to the data")
    print(f"total Line so far is {len(df)}")
    clear_output(wait=True)
    
```

```python
month=input("Please enter month in this format (mmm-yy) eg: for January 2025 it is Jan-25: ")
df['Extracted On']=month 
```

```python
Ofolder=input("paste your output folder here :")
filename=f"Warranty_report_Orielly_{month}"
df.to_csv(Ofolder+f'\\{filename}.csv')
```
