```python
import pandas as pd
import warnings
warnings.filterwarnings("ignore")
```

```python
papath=r"_ProductAttribute.csv"

pallpath=r"HOP20250222_PIESData72.csv"
```

```python
df_pa=pd.read_csv(papath,delimiter="|")

df_pall=pd.read_csv(pallpath,delimiter="|")
```

```python
df_pa
```

```python
df_pall
```

```python
df_merged=df_pa.merge(df_pall,how='left',left_on="PartNumber",right_on="PartNumber")
```

```python
df_merged
```

```python
collist=['DESC_SHO', 'AttributeID' ]
```

```python
df_final=df_merged[collist]
df_final.drop_duplicates().reset_index(drop=True)
```

```python
OFolder=r"Output folder"
```

```python
with pd.ExcelWriter(OFolder+'\\'+f'Hopkins_Parts.xlsx') as writer:  # doctest: +SKIP
    df_pa.to_excel(writer,index=False, sheet_name='Attributes')
    df_pall.to_excel(writer,index=False, sheet_name='Part_Details')
    df_merged.to_excel(writer,index=False, sheet_name='Merged')
```
