```python
df_BPI_customer=pd.read_excel(r"C:\Users\vikram.vadhirajan\OneDrive - Trico\Documents - Product Support India Romania\Catalog\EPR\Sales\BUFiles\BPI_Customers.xlsx")
df_BPI_customer
```

```python
df_merged = pd.merge(df_BPI, df_BPI_customer, left_on='CustomerDesc', right_on='Customer', how='left')
```

```python
df_merged
```

```python
Customers=df_merged["Customer Cleaned"].unique().tolist()
```

```python
df_merged=df_merged[['BusinessUnit', 'OrganizationDesc', 'CustomerCode', 'Customer Cleaned', 'CustomerDesc',
       'CustomerAddress1', 'CustomerState', 'CustomerZip', 'ProductCode',
       'ProductDesc', 'QTY SHIPPED']]
```

```python
FileName=r"C:\Users\vikram.vadhirajan\OneDrive - Trico\Documents - Product Support India Romania\Catalog\EPR\Sales\BUFiles\BPI_ASC_IBI_CustomerWise_Sales.xlsx"
with pd.ExcelWriter(FileName, engine='xlsxwriter') as writer:
    for customer in Customers:
        df_customer = df_merged[df_merged["Customer Cleaned"] == customer]
        if df_customer.empty:
            print(f"No data for {customer}, skipping.")
            continue
        print(f"Saved {customer} with {len(df_customer)} rows.")
        df_customer.to_excel(writer, sheet_name=str(customer), index=False)
```
