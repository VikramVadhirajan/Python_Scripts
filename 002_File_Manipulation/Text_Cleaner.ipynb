```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from time import sleep
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.support import expected_conditions as EC
import pandas as pd
import datetime
import warnings


# Ignore all warnings (not recommended in general)
warnings.filterwarnings("ignore")
```

```python

f=open(r"file.txt",'r')
Linebraker='marketplace-product row '
```

```python
content=f.read()
List=content.split(Linebraker)
print(len(List))
```

```python
List[1]
```

```python
prefix="https://parts.nissanusa.com"
```

```python
List[1].split('title="')[2].split('"')[0]
```

```python
List[1].split('title="')[1].split(" : ")[1].split(":")[0]
```

```python
cols =['Sl.No'] #,'Review_Mentions'
df = pd.DataFrame(columns=cols)
count=0
```

```python
for i in range (1,len(List)):
    df.loc[count,'Sl.No']=i
    df.loc[count,'Name']=List[i].split('title="')[2].split('"')[0]
    df.loc[count,'PartNumber']=(List[i].split('title="')[1].split(" ")[0])
    df.loc[count,'MSRP']=(float(List[i].split(' MSRP </span>')[1].split('</div>')[0].strip().replace("$","").replace(",","")))
    df.loc[count,'Fitment']=(List[i].split('fitment-model ">')[1].split('</li>')[0])
    try:
        df.loc[count,'Replaces']=(List[i].split('Replaces:</strong>\n\t')[1].split('</div>')[0])
    except:
        df.loc[count,'Replaces']=('null')
    df.loc[count,"Link"]=prefix+List[i].split('href="')[1].split('?')[0]
    count=count+1
```
