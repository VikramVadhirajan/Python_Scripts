```python
#-----------------------------------------------------------------------------------------------------------------
df3['Model_Years']=df3['Model_Years'].str.split(',')
df3=df3.explode('Model_Years')
df3 = df3.drop_duplicates()
df3 = df3.reset_index(drop=True)


#-----------------------------------------------------------------------------------------------------------------
for i in range(len(df3)):
    if "-" in df3['Model_Years'][i]:
        #print(i)
        year=df3['Model_Years'].str.split('-')[i]
        startyear=year[0]
        endyear=year[1]
        df3.loc[i,'Start_Year']=startyear
        df3.loc[i,'End_Year']=endyear
    if "-" not in df3['Model_Years'][i]:
        #print (i)
        df3.loc[i,'Start_Year']=df3['Model_Years'][i]
        df3.loc[i,'End_Year']=df3['Model_Years'][i]
#-----------------------------------------------------------------------------------------------------------------
df3['Start_Year']=df3['Start_Year'].str.replace(" ","").astype(int)
df3['End_Year']=df3['End_Year'].str.replace(" ","").astype(int)

#-----------------------------------------------------------------------------------------------------------------
del df3['Model_Years']
#-----------------------------------------------------------------------------------------------------------------
```

```python
df3['Years'] = df3.apply(lambda row: list(range(row['Start_Year'], row['End_Year'] + 1)), axis=1)
df3.drop(columns={'End_Year','Start_Year'})
df3=df3.explode('Years')
df3=df3.drop(columns={'End_Year','Start_Year'})
```

```python
cols1=["Part Number","Attribute","Value","Link"]
cols2=["Part Number","Part Name","Category","Attributes","Value","Link"]
cols3=["Part Number","Product_Name","Make","Models","Years","Link"]
df1=df1[cols1]
df2=df2[cols2]
df3=df3[cols3]
```
