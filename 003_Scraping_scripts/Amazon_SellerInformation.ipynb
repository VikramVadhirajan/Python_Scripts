```python
for i in tqdm(range (len(Links))):
    driver.get(Links[i])
    sleep(2)
    df.loc[count, "Link"]=Links[i]
    try:
        df.loc[count, "Title"]=driver.find_elements(By.CLASS_NAME, 'product-title-word-break')[0].text
    except:
        df.loc[count, "Title"]="Not Found"

    try:
        df.loc[count, "Sales Last Month"]=int(driver.find_element(By.ID, "social-proofing-faceout-title-tk_bought").text.split("+")[0].replace('K','000'))
    except:
        df.loc[count, "Sales Last Month"]="Not Found"
    try:
        price=driver.find_element(By.CSS_SELECTOR, '[data-feature-name="corePrice"]').text.replace('''\n''','.').replace('$','').replace(',','')
        if "count"in price:
            percount=float(driver.find_element(By.CSS_SELECTOR, '[data-feature-name="corePrice"]').text.replace('''\n''','.').replace('$','').replace(',','').split(".(")[0])
            #print(percount)
            df.loc[count, "Price"]=percount
        else:
            df.loc[count, "Price"]=float(price)
    except:
        df.loc[count, "Price"]="Not Found"
    try:
        df.loc[count, "ListPrice"]=float(driver.find_element(By.CLASS_NAME, 'a-text-price').text.replace('''\n''','.').replace('$','').replace(',',''))
    except:
        df.loc[count, "ListPrice"]="Not Found"
    try:
        df.loc[count, "Ship_From"]=driver.find_elements(By.CLASS_NAME, 'odf-truncation-popover')[0].text
    except:
        df.loc[count, "Ship_From"]="Not Found"
    try:
        df.loc[count, "Sold_By"]=driver.find_elements(By.CLASS_NAME, 'odf-truncation-popover')[2].text
    except:
        df.loc[count, "Sold_By"]="Not Found"
    try:
        attcount=len(driver.find_elements(By.CLASS_NAME,"prodDetSectionEntry"))
        for i in range(attcount):
            if driver.find_elements(By.CLASS_NAME,"prodDetSectionEntry")[i].text=="Brand":
                df.loc[count, "BrandName"]=driver.find_elements(By.CLASS_NAME,"prodDetAttrValue")[i].text
            elif driver.find_elements(By.CLASS_NAME,"prodDetSectionEntry")[i].text=="Manufacturer":
                df.loc[count, "BrandName"]="Not Found"
                df.loc[count, "Manufacturer"]=driver.find_elements(By.CLASS_NAME,"prodDetAttrValue")[i].text
            else:
                pass
    except:
        pass
    count=count+1
    clear_output(wait=True)
    #print("Completed: ", count, " out of ", len(Links))    
```
