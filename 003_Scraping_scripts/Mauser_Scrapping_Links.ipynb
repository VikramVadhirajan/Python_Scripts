# 4. Defining the Dataframe and Extracting the data into the Dataframe

```python
cols =['Sl.No'] #,'Review_Mentions'
df = pd.DataFrame(columns=cols)
```

```python
count=0
for i in range(1,len(List)):
    moql=List[i].count('data-qty')
    for d in range(1,moql):
        moq=List[i].split('data-qty="')[d].split('''">''')[0]
        #print(moq)
        j=1
        if int(moq) >0:
            df.loc[count,"MoQ"]=moq
            comb=str(i)+'_'+str(j)
            #print(comb)
            df.loc[count,'Price']=List[i].split('lblPrice_'+comb)[1].split('''nowrap">''')[1].split('''</span>''')[0]
            df.loc[count,"Sl.No"]=i
            df.loc[count,"Supplier Name"]=(List[i].split('lnkSupplierPage')[1].split('''/">''')[1].split('''</a>''')[0])
            try:
                df.loc[count,"Stock"]=(List[i].split('available-amount">')[1].split('''</''')[0])
            except:
                df.loc[count,"Stock"]=0
            try:
                df.loc[count,"Leadtime"]=(List[i].split('avail-status">')[1].split('''</''')[0])
            except:
                df.loc[count,"Leadtime"]=0
            try:
                df.loc[count,"Part Number"]=(List[i].split('"mpart-number-lbl">')[1].split('''</span>''')[0])
            except:
                df.loc[count,"Leadtime"]=0
            count=count+1
            j=j+1
        else:
            break
```
