```python
import os
import zipfile

# Define source directory containing ZIP files
zip_dir = r"C:\Users\vikram.vadhirajan\OneDrive - Trico\Documents - Product Support India Romania\Catalog\Score card\Standard\Import files\PDM"  # Replace with your ZIP files folder path

# Define the destination folder where files will be extracted
extract_dir = r"C:\Users\vikram.vadhirajan\OneDrive - Trico\Documents - Product Support India Romania\Catalog\Score card\Standard\Import files\PDM\excel Files"  # Replace with your desired extraction path

# Ensure the extraction directory exists
os.makedirs(extract_dir, exist_ok=True)

# Iterate through each ZIP file in the directory
for file in os.listdir(zip_dir):
    if file.endswith(".zip"):  # Process only .zip files
        zip_path = os.path.join(zip_dir, file)
        
        # Extract ZIP file
        with zipfile.ZipFile(zip_path, 'r') as zip_ref:
            zip_ref.extractall(extract_dir)
        
        print(f"Extracted: {file}")

print("All ZIP files have been extracted successfully!")
```
