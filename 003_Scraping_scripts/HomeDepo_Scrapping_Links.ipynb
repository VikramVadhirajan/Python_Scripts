# 4. Defining the Dataframe and Extracting the data into the Dataframe

```python
Cols=["Links","Name"]
df = pd.DataFrame(columns=Cols)
df
count=0
```

```python

for i in range(10):
    driver.execute_script("window.scrollTo(0, 4600);")
    sleep(1)
    product_pods = driver.find_elements(By.CSS_SELECTOR, '[data-testid="product-header"][data-component="ProductHeader"]')
    print(len(product_pods))
    # Create a list to store product information
    product_info_list = []
    row=0
    # Iterate through each product pod
    for product_pod in product_pods:
        if row<24:
            # Extract product title
            product_title = product_pod.find_element(By.CLASS_NAME, 'product-header__title-product--ey0fs').text.strip()

            # Extract href value
            product_href = product_pod.find_element(By.CSS_SELECTOR, 'a').get_attribute('href').strip()

            # Append the product information to the list
            df.loc[count,'Links']=product_href
            df.loc[count,'Name']=product_title
            row=row+1
            count=count+1
        else:
            break

    elemento = driver.find_element(By.CLASS_NAME,"results-pagination__counts")
    print(elemento.text)
    # Print the text content of the element
    #print("Element Text:", element.text)
    max=elemento.text.split('of')[0].split('-')[1]
    TotalPage=elemento.text.split('of')[1].split('results')[0]
    print(max,TotalPage)

    if max!= TotalPage:
        next_button = driver.find_element(By.CSS_SELECTOR, '[aria-label="Next"]')
        #driver.execute_script("window.scrollTo(0, 3000);")
        #driver.execute_script("arguments[0].scrollIntoView();", next_button)
        next_button.click()
        sleep(1)
        #driver.refresh()
        i=i+1
        sleep(2)
        print("---------------------------------")
    else:
        break
```
