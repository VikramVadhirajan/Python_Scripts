# 4. Defining the Dataframe and Extracting the data into the Dataframe

```python
Cols=["Links","Name"]
df = pd.DataFrame(columns=Cols)
```

```python
driver.find_element(By.CSS_SELECTOR,'[class="next-page"]').click()
```

```python
count=0
for i in range(3):
        elements = driver.find_elements(By.CSS_SELECTOR, 'div.columns.medium-17.large-18.show-for-medium h2 a')

        # Extract href and title from each element
        for element in elements:
            href = element.get_attribute('href')
            title = element.text.strip()
            df.loc[count,'Sl.No']=count+1
            df.loc[count,'Links']=href
            df.loc[count,'Name']=title
            count=count+1        
        driver.find_element(By.TAG_NAME,'body').send_keys(Keys.END)
        sleep(1)
        next_button = driver.find_element(By.CSS_SELECTOR,'[class="next-page"]')
        next_button.click()
        i=i+1
        print("No of pages =", i)
```
