```python
cols=["Sl.No"]
df3= pd.DataFrame(columns=cols)
count3=0
```

```python
for i in tqdm(range(97,len(List))):
    print(i)
    driver.get(home)
    sleep(1)
    typebox=wait.until(EC.presence_of_element_located((By.ID,"CreateVehicle_suggest_box")))
    sleep(1)
    typebox.send_keys(List[i])
    wait.until(EC.presence_of_all_elements_located((By.CLASS_NAME,"ui-menu-item")))[0].click()
    try:
        wait.until(EC.element_to_be_clickable((By.CSS_SELECTOR,'[id="linkSeeMore"]'))).click()
    except:
        continue
    sleep(10)
    wait.until(EC.element_to_be_clickable((By.CSS_SELECTOR,'[id="tb-search-box"]'))).click()
    searchbox=wait.until(EC.element_to_be_clickable((By.CSS_SELECTOR,'[id="tb-search-box"]')))
    searchbox.send_keys("Wiper Arm(s)")
    wait.until(EC.element_to_be_clickable((By.CSS_SELECTOR,'[id="bt-search"]'))).click()
    sleep(2)
    for exp in range (100):
        try:
            WebDriverWait(driver, 2).until(EC.presence_of_element_located((By.CLASS_NAME,'dynatree-ico-cf'))).click()
        except:
            break
    sbs=wait.until(EC.presence_of_all_elements_located((By.CSS_SELECTOR,f'[title="{"WIPER ARM"}"]')))
    for w in range(len(sbs)):
        sbs[w].click()
        sleep(1)
    es=WebDriverWait(driver, 15).until(EC.presence_of_all_elements_located((By.CSS_SELECTOR,'[class="part-labor-table"]')))
    for e in es:
        if 'Mfg.' in e.text:
            if "$" in e.text:
                pns=e.text.count("$")
                #print(pns)
                for pn in range(pns):
                    Appnote=e.text.split("$")[pn].split("\n")[-2]
                    PN=e.text.split("$")[pn].split("\n")[-1]
                    Price=float(e.text.split("$")[pn+1].split("\n")[0])
                    print(Appnote,"|",PN,"|",Price)
                    df3.loc[count3,"Sl.No"]=""
                    df3.loc[count3,"Keys"]=""
                    df3.loc[count3,"Links"]=""
                    df3.loc[count3,"Appln Notes"]=Appnote
                    df3.loc[count3,"OE Part Number"]=PN
                    df3.loc[count3,"Price [$]"]=Price
                    count3=count3+1
            else:
                df3.loc[count3,"Sl.No"]=""
                df3.loc[count3,"Keys"]=List[i]
                df3.loc[count3,"Links"]=driver.current_url
                df3.loc[count3,"OE Part Number"]="['"+e.text.split('\n')[4]+"']"
                count3=count3+1
        else:
            df3.loc[count3,"Sl.No"]=""
            df3.loc[count3,"Keys"]=List[i]
            df3.loc[count3,"Links"]=driver.current_url
            df3.loc[count3,"OE Part Number"]="No Part Number"
            count3=count3+1
    driver.find_element(By.CSS_SELECTOR,'[title="New Vehicle"]').click()
    clear_output(True)
```
