```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from time import sleep
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from urllib.parse import unquote
from selenium.webdriver.support.ui import Select
import pandas as pd
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.common.action_chains import ActionChains
import datetime 
import warnings
# Ignore all warnings (not recommended in general)
warnings.filterwarnings("ignore")
from IPython.display import clear_output 
```

```python
path= 'C://chromedriver.exe'
driver=webdriver.Chrome()
driver.maximize_window()
wait=WebDriverWait(driver, 10)
driver.get("https://www.gmpartsgiant.com/")
```

```python
cols=["Sl.No"]
df= pd.DataFrame(columns=cols)
count=0
```

```python
List=["19351570",
"19351572",
"19351591",
"12192462",
"12192470",
"19301298",
"19301299",
"19170842",
"19171855",
"19180531",
"12192471",
"12192375",
"19170849",
"19351593",
"19171849"
]
print(len(List))
```

```python
from tqdm import tqdm
for i in tqdm(range(len(List))): #len(List)):
    print(f"running {i}th element | {List[i]}")
    search_text=f"{List[i]}"
    searchbox=wait.until(EC.element_to_be_clickable((By.CLASS_NAME,"ab-input-control")))
    sleep(1)
    searchbox.click()
    searchbox.send_keys(search_text)
    searchbox.submit()
    sleep(1)
    element=wait.until(EC.presence_of_element_located((By.XPATH, ".//div[@class='pn-tab-title'][h2/text()='Vehicle Fitment']")))
    driver.execute_script("arguments[0].scrollIntoView();", element)
    element.click()
    try:
        WebDriverWait(driver, 3).until(EC.presence_of_element_located((By.CLASS_NAME,"fit-vehicle-list-view-text"))).click()
    except:
        pass
    tablecontent=driver.find_elements(By.CLASS_NAME,"fit-vehicle-list-text")

    for i in range(len(tablecontent)):
        if i%3==0:
            df.loc[count,"YMM"]= tablecontent[i].text
        if i%3==1:
            df.loc[count,"EnineList"]= tablecontent[i].text
        if i%3==2:
            df.loc[count,"Vehicle Options"]= tablecontent[i].text
            df.loc[count,"PartNumber"]=search_text
            count=count+1
```

```python
for i in range(len(df)):
    if "," in df["YMM"][i]:
        n=df["YMM"][i].count(",")
        df.at[i,"Model_Years"]=df['YMM'][i].split(" ")[0:(1+n)]
        df.loc[i,"Make"]=df["YMM"][i].split(" ")[(1+n)]
        df.loc[i,"Model"]=" ".join(df["YMM"][i].split(" ")[(2+n):])        
        print(n)
    else:
        df.at[i,"Model_Years"]=df['YMM'][i].split(" ")[0:(1)]
        df.loc[i,"Make"]=df["YMM"][i].split(" ")[1]
        df.loc[i,"Model"]=" ".join(df["YMM"][i].split(" ")[2:])

df=df.explode("Model_Years")

df = df.reset_index(drop=True)
for i in range(len(df)):
    if "-" in df['Model_Years'][i]:
        #print(i)
        year=df['Model_Years'].str.split('-')[i]
        startyear=year[0]
        endyear=year[1]
        df.loc[i,'Start_Year']=startyear
        df.loc[i,'End_Year']=endyear
    if "-" not in df['Model_Years'][i]:
        #print (i)
        df.loc[i,'Start_Year']=df['Model_Years'][i]
        df.loc[i,'End_Year']=df['Model_Years'][i]


df['Start_Year']=df['Start_Year'].str.replace(",","").replace(" ","").astype(int)
df['End_Year']=df['End_Year'].str.replace(",","").replace(" ","").astype(int)
```

```python
df
```

```python
df['Years'] = df.apply(lambda row: list(range(row['Start_Year'], row['End_Year'] + 1)), axis=1)
df.drop(columns={'End_Year','Start_Year'})
df=df.explode('Years')
df=df.drop(columns={'End_Year','Start_Year'})

df["Engine"]=df["EnineList"].str.split(", ")
df=df.explode("Engine")
```

```python
from tqdm import tqdm
for i in tqdm(range(10,len(List))): #len(List)):
    print(f"running {i}th element | {List[i]}")
    search_text=f"{List[i]}"
    searchbox=wait.until(EC.element_to_be_clickable((By.CLASS_NAME,"ab-input-control")))
    sleep(1)
    searchbox.click()
    searchbox.send_keys(search_text)
    searchbox.submit()
    try:
        wait.until(EC.presence_of_element_located((By.CLASS_NAME,"ab-select-col"))).find_element(By.XPATH,".//li[1]").click()
    except:
        pass
    sleep(1)
    result=wait.until(EC.presence_of_element_located((By.CLASS_NAME,"ns-main-title"))).text
    if "Search Results" in result:

        Names=wait.until(EC.visibility_of_all_elements_located((By.CSS_SELECTOR,'[class="ns-parts-list-desc"]')))
        PartNumbers=driver.find_elements(By.CSS_SELECTOR,'[class="ns-parts-list-parts"]')
        for name, pno in zip(Names, PartNumbers):
            App=driver.find_element(By.CLASS_NAME,"av-side-rqi").text
            try:
                Spec=driver.find_element(By.CLASS_NAME,"av-side-opi").text
            except:
                Spec=" "
            df.loc[count,"Application"]= App+" " +Spec
            df.loc[count,"Name"]=name.text
            df.loc[count, "Sl.No"]=i
            df.loc[count,"OrignialSelection"]=List[i]
            df.loc[count,"Number"]=pno.text.split("\n")[0]
            count=count+1
    else:
        print("No results")
    driver.find_element(By.CLASS_NAME,"av-side-change-btn").click()
    sleep(1)
    driver.find_element(By.CLASS_NAME,"av-prev-clear-btn").click()
    sleep(3)
    clear_output(True)   
        
```
