```python
Names=driver.find_elements(By.CLASS_NAME,'product-title')
PartNumbers=driver.find_elements(By.CLASS_NAME,'product-partnum-category')
Values=driver.find_elements(By.CLASS_NAME,'product-prices')
FitmentNotes=driver.find_elements(By.CLASS_NAME,'catalog-product-note')
Links=driver.find_elements(By.XPATH,'//strong[@class="product-title"]/a')
Replaces=driver.find_elements(By.CSS_SELECTOR,'[class="catalog-product-more-in-category catalog-info-row"]')

for Name, PN, Cost,  Link, fitment, Replace in zip(Names, PartNumbers, Values, Links,FitmentNotes,Replaces):
    if "Ignition Cable Package" in Name.text:
        df.loc[count,'Name']=Name.text
        df.loc[count,'Mopar Part Number']=PN.text.split(": ")[1].split(" ")[0]
        df.loc[count,'MSRP']=float(Cost.text.split("$")[1])
        df.loc[count,'Links']=(Link.get_attribute('href').split("?")[0])
        df.loc[count,'Fitment Notes']=fitment.text
        df.loc[count,'Replacement Part Numbers']=Replace.text
        count=count+1
    else:
        pass
```
