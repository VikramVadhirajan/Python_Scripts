# 3. Setting Webdriver and Website Specific Information

```python
content=f.read()
```

```python
List=content.split(Linebraker)
```

```python
print(len(List))
length=len(List)
```

# 4. Defining the Dataframe and Extracting the data into the Dataframe

```python
cols =['Sl.No'] #,'Review_Mentions'
df = pd.DataFrame(columns=cols)
```

```python
count=0

for i in range(1,length):
    df.loc[count,'Sl.No']=i
    #df.loc[count,'Name']=List[i].split('="3"><!--F#f_0-->')[1].split('<!--F/-->')[0]
    df.loc[count,'Current Price']=float(List[i].split('s-item__price')[1].split('</span>')[0].split('$')[1].split('<')[0])
    try:
        df.loc[count,'List Price']=float(List[i].split('STRIKETHROUGH">$')[1].split('</span>')[0])
    except:
        df.loc[count,'List Price']=float(List[i].split('s-item__price')[1].split('</span>')[0].split('$')[1].split('<')[0])
    try:
        df.loc[count,'Rating']=float(List[i].split(' out of 5 stars')[0].rsplit('"clipped">')[-1])
    except:
        pass
    try:
        df.loc[count,'No of Ratings']=int(List[i].split(' product ratings')[0].rsplit('f_0-->')[-1])
    except:
        pass
    try:
        df.loc[count,'Sales_LastMonth']=int(List[i].split(' sold')[0].rsplit('''LD">''')[-1].replace(",",""))
    except:
        pass
    df.loc[count,'Links']=List[i].split('''s-item__link" href="''')[1].split('''?''')[0]
    df.loc[count,'Source']=Source
    df.loc[count,'Product']=search_text
    count=count+1
```

```python
df=df.drop_duplicates(subset=['Links'])
```
