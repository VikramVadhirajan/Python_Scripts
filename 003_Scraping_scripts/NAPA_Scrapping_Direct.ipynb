# 4. Defining the Dataframe and Extracting the data into the Dataframe

```python
cols =['Sl.No'] #,'Review_Mentions'
df = pd.DataFrame(columns=cols)
```

```python
count=0

for i in range(1,len(List)):
    df.loc[count,'Sl.No']=int(i)
    try:
        df.loc[count,'Current Price']=float(List[i].split('data-price="')[1].split('"><div>')[0])
    except:
        pass
    df.loc[count,'Name']=List[i].split('"geo-prod_pod_title">')[1].split('</a><div')[0]
    df.loc[count,'Links']=prefix+List[i].split('href="')[1].split('"')[0]
    df.loc[count,'Source']=Source
    try:
        df.loc[count,'Brand']=List[i].split('data-geo-brand="')[1].split('"')[0]
    except:
        df.loc[count,'Brand']=""
    df.loc[count,'Part Number']=List[i].split('part-number="')[1].split('" class="')[0]
    count=count+1
```
