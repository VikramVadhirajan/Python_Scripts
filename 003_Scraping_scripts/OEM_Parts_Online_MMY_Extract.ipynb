```python
for i in tqdm(range(3977,len(df_in))):
    print(i)
    driver.get(df_in["Link"][i])
    wait.until(EC.presence_of_element_located((By.CLASS_NAME,'product-title')))
    sleep(1)
    try:
        driver.find_element(By.CSS_SELECTOR,'[aria-controls="tab-vehicle-fitment"]').click()
    except:
        try:
            driver.find_elements(By.CLASS_NAME,"accordion-title")[1].click()
        except:
            df.loc[count,"Sl.No"]=count+1
            df.loc[count,"PartNumber"]=df_in["PartNumber"][i]
            df.loc[count,"Link"]=df_in["Link"][i]
            df.loc[count,"Vehicle_Fitment"]="No"
            count=count+1
    
    sleep(1)

    try:
        driver.find_element(By.CLASS_NAME,"fitment-expander").click()
        sleep(1)
    except:
        pass

    VehCount=1

    years=driver.find_elements(By.CLASS_NAME,"fitment-year")
    makes=driver.find_elements(By.CLASS_NAME,"fitment-make")
    models=driver.find_elements(By.CLASS_NAME,"fitment-model")
    trims=driver.find_elements(By.CLASS_NAME,"fitment-trim")
    engines=driver.find_elements(By.CLASS_NAME,"fitment-engine")
    try:
        status=driver.find_element(By.CLASS_NAME,"message-title ").text
    except:
        status=""

    for year, make, model, trim, engine in zip( years[1:], makes[1:], models[1:], trims[1:], engines[1:]):
        df.loc[count,"Sl.No"]=count+1
        df.loc[count,"Status"]=status
        df.loc[count,"VehicleCounts"]=VehCount
        df.loc[count,"PartNumber"]=df_in["PartNumber"][i]
        df.loc[count,"Year"]=year.text
        df.loc[count,"Make"]=make.text
        df.loc[count,"Model"]=model.text
        df.loc[count,"Body & Trim"]=trim.text
        df.loc[count,"Engine & Transmission"]=engine.text
        df.loc[count,"Vehicle_Fitment"]="Yes"
        df.loc[count,"Link"]=df_in["Link"][i]
        try:
            rep=driver.find_element(By.CLASS_NAME,"product-superseded-list").text.split(": ")[1]
            df.loc[count,"Replacement"]=rep
        except:
            df.loc[count,"Replacement"]=""
        try:
            sts=driver.find_elements(By.CLASS_NAME,"product-subtitle")
            df.loc[count,"Range"]=sts[len(sts)-1].text
        except:
            pass
        try:
            df.loc[count,"Position"]=driver.find_element(By.CSS_SELECTOR,'[class="positions"]').text.split('Positions: ')[1]
        except:
            pass
        try:
            df.loc[count,"OtherNames"]=driver.find_element(By.CSS_SELECTOR,'[class="also_known_as"]').text.split('Other Names: ')[1]
        except:
            pass
        count=count+1
        VehCount=VehCount+1
    clear_output(wait=True)
```
