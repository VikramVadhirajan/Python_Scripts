```python
cols1 =['Sl.No'] #,'Review_Mentions'
df1= pd.DataFrame(columns=cols1)
```

```python
count=0

for j in range(2):
    print(cols[j])
    for i in range(len(df)):
        if "-------" not in str(df.loc[i,cols[j]]):
            df1.loc[count,"Sl.No"]=count+1
            df1.loc[count,"Part Number Type"]=cols[j]        
            df1.loc[count,"Part Number"]=str(df.loc[i,cols[j]])
            count=count+1
```

```python
path= 'C://chromedriver.exe'
driver=webdriver.Chrome()
wait=WebDriverWait(driver, 5)
driver.get('https://apdi.mycarparts.net/home')
```

```python
cols2 =['Sl.No'] #,'Review_Mentions'
df2= pd.DataFrame(columns=cols2)
count=0
```

```python
from tqdm import tqdm
for i in tqdm(range(2050,len(df1))):
    print(i)
    search_bar=driver.find_element(By.CSS_SELECTOR,'[placeholder="Part Number"]')
    search_bar.click()
    p=df1['Part Number'][i]
    search_bar.send_keys(p)
    search_bar.submit()
    sleep(1)
    es=driver.find_elements(By.CSS_SELECTOR,'[style="border-left: 0; border-right: 0"]')
    ds=driver.find_elements(By.CSS_SELECTOR,'[class="link"]')
    for e,d in zip(es,ds):
        df2.loc[count,"Sl.No"]=count
        df2.loc[count,"Agility Interchange"]=e.text.split('\n')[0]
        df2.loc[count,"Agility Product"]=e.text.split('\n')[1]
        df2.loc[count,"Part Number Type"]=df1.loc[i,"Part Number Type"]     
        df2.loc[count,"Part Number"]=df1.loc[i,"Part Number"]
        df2.loc[count,"index1"]=i
        df2.loc[count,"Interchange_BrandPartNumber"]=d.text.split("\n")[0]
        df2.loc[count,"Brand"]=d.text.split("\n")[1].split(": ")[1]
        count=count+1
    search_bar.clear()
    clear_output(wait=True)
```

```python
colls=['Sl.No','index1','Part Number Type','Part Number',"Interchange_BrandPartNumber",'Brand','Agility Product','Agility Interchange']
df2=df2[colls]
```
