```python
for l in tqdm(range(818,len(List))):
    print(l,List[l])
    searchbox=driver.find_element(By.ID,"search")
    searchbox.click()
    searchbox.send_keys(List[l])
    searchbox.send_keys(Keys.ENTER)
    sleep(1)
    try:
        try:
            driver.find_elements(By.CLASS_NAME,"product-item-link")[0].click()
        except:
            pass
        left=driver.find_elements(By.CLASS_NAME,'product-basic-details__text--left')
        right=driver.find_elements(By.CLASS_NAME,'product-basic-details__text--right')
        try:
            PartName=driver.find_element(By.CLASS_NAME,'product-basic-details__text--title').text
            for i, j in zip( left, right):
                df1.loc[count1,'PartNumber']=List[l]
                df1.loc[count1,'PartName']=PartName 
                df1.loc[count1,"Attribute"]=i.text
                df1.loc[count1,"Value"]=j.text
                count1=count1+1
            try:        
                df2.loc[count2,'PartNumber']=List[l]
                df2.loc[count2,'PartName']=PartName   
                df2.loc[count2,"Application"]=driver.find_element(By.CLASS_NAME,"app-summary--text").text
                count2=count2+1
                clear_output(wait=True)   
            except:
                pass
            driver.find_element(By.CLASS_NAME,'div-return-plp').click()
        except:
            searchbox=driver.find_element(By.ID,"search")
            searchbox.clear()
    except:
        searchbox=driver.find_element(By.ID,"search")
        searchbox.clear()
        pass
    sleep(1)
```
