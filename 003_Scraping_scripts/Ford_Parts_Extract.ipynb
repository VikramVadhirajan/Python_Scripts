```python
path= 'C://chromedriver.exe'
driver=webdriver.Chrome()
driver.get('https://parts.ford.com/shop/en/us/engine/spark-plugs#facet:&productBeginIndex:0&orderBy:&pageView:&minPrice:&maxPrice:&pageSize:&')
```

```python
driver.find_elements(By.CLASS_NAME,"pNameWidth")[1].text
```

```python
driver.find_elements(By.CLASS_NAME,"pNumber")[1].text.split(":")[1]
```

```python
driver.find_elements(By.CSS_SELECTOR,'[class="partMSRP hidden-xs"]')[1].text.split("$")[1]
```

```python
driver.find_elements(By.XPATH,'//div[@class="partImageContainer"]/a')[1].get_attribute('href')
```

```python
Cols=["Links","Name"]
df = pd.DataFrame(columns=Cols)
count=0
```

```python
names=driver.find_elements(By.CLASS_NAME,"pNameWidth")
Part_Numbers=driver.find_elements(By.CLASS_NAME,"pNumber")
MSRPS=driver.find_elements(By.CSS_SELECTOR,'[class="partMSRP hidden-xs"]')
Links=driver.find_elements(By.XPATH,'//div[@class="partImageContainer"]/a')

for name,pn,msrp,link in zip(names,Part_Numbers,MSRPS,Links):
    df.loc[count,"Name"]=name.text
    df.loc[count,"PartNumber"]=pn.text.split(":")[1]
    df.loc[count,"MSRP"]=float(msrp.text.split("$")[1])
    df.loc[count,"Links"]=link.get_attribute('href')
    count=count+1

len(df)
```
