# 3. Setting Webdriver and Website Specific Information

```python
content=f.read()
List=content.split(Linebraker)
print(len(List))
```

# 4. Defining the Dataframe and Extracting the data into the Dataframe

```python
cols =['Sl.No'] #,'Review_Mentions'
df = pd.DataFrame(columns=cols)
```

```python
count=0

for i in range(1,len(List)):
    df.loc[count,'Sl.No']=i
    try:
        df.loc[count,'Current Price']=float(List[i].split('"az_UPc">')[1].split('</')[0]+"."+List[i].split('"az_QPc">')[1].split('</')[0])
    except:
        df.loc[count,'Current Price']=0
    try:
        df.loc[count,'Rating']=float(List[i].split('bv_text">')[1].split ('</div>')[0])
    except:
        df.loc[count,'Rating']=0
    try:
        df.loc[count,'No of Ratings']=int(List[i].split('bv_text">(')[1].split (')</div>')[0])
    except:
        df.loc[count,'No of Ratings']=0
    df.loc[count,'Name']=List[i].split('-product-title">')[1].split ('</h3')[0]
    df.loc[count,'Product']=search_text
    df.loc[count,'Links']=prefix+List[i].split(' href="')[1].split ('?searchText')[0]
    df.loc[count,'Part Number']=List[i].split('"az_LTb"> ')[1].split ('</span>')[0]
    #df.loc[count,'ASIN']=List[i].split('"az_XMb"> ')[2].split ('</span>')[0]
    df.loc[count,'Source']=Source
    count=count+1
```
