# 3. Setting Webdriver and Website Specific Information

```python
path= 'C://chromedriver.exe'
driver=webdriver.Chrome()
driver.get('https://www.go-parts.com/')
driver.maximize_window()
wait=WebDriverWait(driver, 5)
```

```python
# Find the search bar by its ID
try:
    search_bar = driver.find_element(By.ID, 'top-search-input')
except:
    pass

# Enter the search text

search_bar.send_keys(search_text)
```

# 4. Defining the Dataframe and Extracting the data into the Dataframe

```python
Cols=["Links","Name"]
df = pd.DataFrame(columns=Cols)
count=0
```

```python
for i in range(50):
    try:
        wait.until(EC.presence_of_element_located((By.CSS_SELECTOR, '[class="ais-infinite-hits--showmore"]'))).click()
        sleep(1)
    except:
        break
```

```python
count=0
nelements=driver.find_elements(By.CSS_SELECTOR,'div.product-info a.product-name')
for element in nelements:
    df.loc[count,'Name']=element.text
    df.loc[count,'Links']=element.get_attribute('href').split("?query")[0]
    count=count+1
```
