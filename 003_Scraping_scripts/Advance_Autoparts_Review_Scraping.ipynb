```python
cols_comment =[ "Part Number", "Product Name", "Title", "Comments","Star", "Review Date"] #,'Review_Mentions','Standards'
df = pd.DataFrame(columns=cols_comment)
count=0
```

```python
numbers=["18-180",
"26-160",
"55-121",
"XC22",
"XC26"
]
```

```python
for s in range(len(numbers)):
    driver.get("https://shop.advanceautoparts.com/")
    searchbox=driver.find_element(By.ID,"search-input")
    searchbox.click()
    searchbox.send_keys(numbers[s])
    searchbox.send_keys(Keys.ENTER)
    sleep(4)
    driver.find_elements(By.CLASS_NAME,"css-141bbgw")[0].find_element(By.CLASS_NAME,'css-1bwgs0x').click()
    sleep(4)
    print(driver.find_element(By.CLASS_NAME,"css-tvrt7a").text)
    try:
        for j in range(500): 
            print (j)
            check1=driver.find_elements(By.CLASS_NAME,"css-50kmc7")[-1].text
            for i in range(5):   
                df.loc[count,"Part Number"]= driver.find_element(By.CLASS_NAME,"css-tvrt7a").text
                df.loc[count,"Product Name"]= driver.find_element(By.CLASS_NAME,"css-1vorsr4").text
                df.loc[count,"Star"]= int(driver.find_elements(By.XPATH, '//div[contains(@class, "css-1nqas9f")]//span[@aria-label]')[i].get_attribute('aria-label').split(" out")[0])
                df.loc[count,"Title"]= driver.find_elements(By.CSS_SELECTOR,'[class="css-1nqas9f"]')[i].find_element(By.CSS_SELECTOR,'[class="css-sske0x"]').text
                df.loc[count,"Review Date"]= driver.find_elements(By.CSS_SELECTOR,'[class="css-1nqas9f"]')[i].find_element(By.CSS_SELECTOR,'[class="css-1fco2k9"]').text
                df.loc[count,"Comments"]= driver.find_elements(By.CSS_SELECTOR,'[class="css-1nqas9f"]')[i].find_element(By.CSS_SELECTOR,'[class="css-uw7cx0"]').text
                count=count+1
            element = driver.find_element(By.CSS_SELECTOR,'[class="css-9dfio9"]')
            actions = ActionChains(driver)
            actions.move_to_element(element).perform()
            element.click()
            sleep(2)
            check2=driver.find_elements(By.CLASS_NAME,"css-50kmc7")[-1].text
            if check1==check2:
                break
        clear_output(wait=True)
    except:
        pass
        clear_output(wait=True)
    driver.get("https://shop.advanceautoparts.com/")
    sleep(1)
```

```python
cols =['Sl.No'] #,'Review_Mentions','Standards'
df = pd.DataFrame(columns=cols)
count=0
```

```python
dols=driver.find_elements(By.CLASS_NAME,"css-15cggpf")
cents=driver.find_elements(By.CLASS_NAME,"css-2izyy6")
titles=driver.find_elements(By.CLASS_NAME,"css-cm8roc")
PNS=driver.find_elements(By.CLASS_NAME,"css-scvhiv")
Links=driver.find_elements(By.CSS_SELECTOR,'div.css-195wmup a.css-vurnku')
for dol,cent,title,PN,Link in zip(dols,cents,titles,PNS,Links):

    df.loc[count,"Sl.No"]=count
    df.loc[count,'Price']=float(dol.text+cent.text)
    df.loc[count,'PartNumber']=PN.text.split("#")[1]
    df.loc[count,"Title"]=title.get_attribute('title')
    df.loc[count,'Link']=Link.get_attribute('href')
    count=count+1

len(df)
```
