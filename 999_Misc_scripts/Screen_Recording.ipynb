```python
import pyscreenrec
import time

# Create a ScreenRecorder instance
recorder = pyscreenrec.ScreenRecorder()
```

```python
time.sleep(3)
# Start recording (output file, FPS, optional monitor settings)
# The start_recording method runs in a background thread
recorder.start_recording("my_recording.mp4", 30) 

print("Recording started for 10 seconds...")
# Record for a specific duration
time.sleep(10) 

# Stop recording and save the file
recorder.stop_recording()

print("Recording saved to my_recording.mp4")
```
