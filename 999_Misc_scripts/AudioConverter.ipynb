```python

def convert_ogg_to_mp3(input_file, output_file):
    """
    Convert an OGG audio file to MP3 format.

    Parameters:
    input_file (str): Path to the input OGG file.
    output_file (str): Path to the output MP3 file.
    """
    # Load the OGG file
    audio = AudioSegment.from_ogg(input_file)

    # Export as MP3 with a specified bitrate (optional, e.g., 192k)
    audio.export(output_file, format="mp3", bitrate="192k")

    print(f"Successfully converted {input_file} to {output_file}")
```

```python
convert_ogg_to_mp3(input_file, output_file)
```

```python
mp3_path = r"C:\Users\vikram.vadhirajan\Downloads\Dashavatara_Stuthi.mp3"
output_path = r"C:\Users\vikram.vadhirajan\Downloads\Dashavatara_Stuthi.mp4"
image_path= r"C:\Users\vikram.vadhirajan\Downloads\Dashavatara_Stuthi_Image.png"
```

```python
def mp3_to_mp4(mp3_path, image_path, output_path):
    # Load the audio file
    audio_clip = AudioFileClip(mp3_path)
    
    # Load the image and set its duration to match the audio
    video_clip = ImageClip(image_path).set_duration(audio_clip.duration)
    
    # Set the audio of the video clip
    video_clip = video_clip.set_audio(audio_clip)
    
    # Write the final result to a video file
    # fps=24 is standard, but you can use fps=1 to save file size for static images
    video_clip.write_videofile(output_path, fps=24, codec="libx264", audio_codec="aac")
    
    # Close clips to release resources
    audio_clip.close()
    video_clip.close()
```

```python

# Example usage
mp3_to_mp4(mp3_path, image_path, output_path)
```
