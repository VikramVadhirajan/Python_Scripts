```python
import pandas as pd
```

```python
df_aa=pd.read_excel(f"D:\99_Downloads\Keyfile_TMP.xlsx",sheet_name="Sheet1",usecols="A:D")
df_bb=pd.read_excel(f"D:\99_Downloads\Keyfile_TMP.xlsx", sheet_name="Sheet2",usecols="A:D")
```

```python
df_bb.head()
```

```python
df_aa["Match"]=None
df_aa
```

```python
for i in range (len(df_aa)):
    best_key = None
    best_score = float("inf")

    for j in range(len(df_bb)):
        if (
            df_bb.loc[j, "Length"] <= df_aa.loc[i, "Length"] and
            df_bb.loc[j, "Width"]  <= df_aa.loc[i, "Width"] and
            df_bb.loc[j, "Height"] <= df_aa.loc[i, "Height"]
        ):

            score = (
                (df_aa.loc[i, "Length"] - df_bb.loc[j, "Length"]) +
                (df_aa.loc[i, "Width"]  - df_bb.loc[j, "Width"]) +
                (df_aa.loc[i, "Height"] - df_bb.loc[j, "Height"])
            )

            if score < best_score:
                best_score = score
                best_key = df_bb.loc[j, "Key"]

    df_aa.at[i, "Match"] = best_key
```
