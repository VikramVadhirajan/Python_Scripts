```python
df1['Name_Lower'] = df1['Name'].str.lower()
df1["Name_Lower"][0]
```

```python
df1['EL']=df1['Name_Lower'].str.replace('inches', ' in ',).str.replace('inch', ' in ',).str.replace('"', " in ").str.split().str.join(" ")
df1["Extended_Length"]=df1["EL"].str.split(" in ",expand=True)[0].apply(lambda x: x.split(' ')[-1]).str.strip()
del df1['EL']
df1[["Name","Extended_Length"]]
```

```python
df1['F']=df1['Name_Lower'].str.replace('lbs', 'lb',).str.replace('pound', 'lb',).str.replace('"', "in").str.replace('/', " / ").str.replace("lb"," lb").str.split().str.join(" ")
df1["Force"]=df1["F"].str.split(" lb ",expand=True)[0].apply(lambda x: x.split(' ')[-1]).str.strip()
del df1['F']
df1[["Name","Force","Extended_Length"]]
```

```python
df1['Extended_Length']=pd.to_numeric(df1['Extended_Length'], errors='coerce')
df1['Force']=pd.to_numeric(df1['Force'], errors='coerce')
```
