## Import Essential libraries

```python
import googlemaps
import json
from datetime import datetime
import pandas as pd
import os
from dotenv import load_dotenv
from time import sleep
```

## Load the API keys

```python
load_dotenv("keys.env")
APIKEY=os.environ.get("geoapikey")
```

## Import and Clean the Data

```python
file=r"C:\13_Tableau\001_BMRCL_Dataset\BMRCL_Station_Name_Updated.csv"
df=pd.read_csv(file)
df['Station Name']=df['Station']+" Metro Station, Bengaluru, Karnataka"
df
```

## Initialize the Client

```python
# Initialize the client
gmaps = googlemaps.Client(key=APIKEY)
```

```python
for i in range (len(df)):

    place_name=df.loc[i,"Station Name"]
    geocode_result = gmaps.geocode(place_name)
    if geocode_result:
        # The first result is usually the most relevant match
        location = geocode_result[0]['geometry']['location']
        df.loc[i,"Latitude"] = location['lat']
        df.loc[i,"Longitude"] = location['lng']
        df.loc[i,"Address"] = geocode_result[0]['formatted_address']
    else:
        print("Location not found.")
    sleep(1)

    # print(place_name)
```

## Write Back the data. 

```python
df
```

```python
df.to_csv(file)
```
