```python
for i in range(len(Plist)):
    search=driver.find_element(By.CSS_SELECTOR,'[id="prt-search-by-input-field"]')
    search.clear()
    search.clear()
    sleep(1)

    search.send_keys(Plist[i])
    search.submit()
    sleep(4)
    try: 
        t=driver.find_element(By.XPATH, "//img[@class='overlaid-image stat-image-link']")
        s=t.get_attribute('src')
        import requests

        # Download the image using requests
        response = requests.get(s, stream=True)

        if response.status_code == 200:
        # Get filename from URL or generate a unique one
            filename=driver.find_element(By.CSS_SELECTOR,'[class="gb-label4 prt-cart-part-number-label-styling"]').text.split(" ")[-1]+".jpg" # Extract filename from URL (modify if needed)
            with open(filename, 'wb') as f:
                for chunk in response.iter_content(1024):
                    f.write(chunk)
            print(f"Image downloaded: {filename}")
        else:
            print(f"Failed to download image: {s}")
            search.clear()
    except:
        print(f"part number not available for {Plist[i]}")

    sleep(1)
```
