## Video Speedup

```python
import cv2
```

```python
inputvideo_path=r"C:\12_Portfolio_Website\videos\demo.mp4"
fastvideo_path=r"C:\12_Portfolio_Website\videos\demo_output.mp4"
```

```python
cap = cv2.VideoCapture(inputvideo_path)
original_fps = cap.get(cv2.CAP_PROP_FPS)
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
```

```python
# Set new speed (e.g., 2x faster)
new_fps = original_fps * 2
fourcc = cv2.VideoWriter_fourcc(*'mp4v')
out = cv2.VideoWriter(fastvideo_path, fourcc, new_fps, (width, height))

while cap.isOpened():
    ret, frame = cap.read()
    if not ret: break
    out.write(frame)

cap.release()
out.release()
```

## Video Trimming

```python
from moviepy.editor import VideoFileClip, concatenate_videoclips
```

```python
inputvideo_path=r"C:\12_Portfolio_Website\videos\demo.mp4"
trimmedvideo_path=r"C:\12_Portfolio_Website\videos\demo_trimmed.mp4"
```

```python

# Load the original video
clip = VideoFileClip(inputvideo_path)

# Cut the video between start_time and end_time (in seconds)
# You can also use strings like "00:00:10"
clip1 = clip.subclip("00:00:00","00:00:37")

clip2 = clip.subclip("00:00:58","00:01:09")

# Combine clips
final_clip = concatenate_videoclips([clip1, clip2])

# Export video
final_clip.write_videofile(trimmedvideo_path, codec="libx264", audio_codec="aac")
```
