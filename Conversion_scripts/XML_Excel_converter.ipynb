```python

def xml_to_xlsx(xml_file, xlsx_file):
    # Parse the XML file
    tree = ET.parse(xml_file)
    root = tree.getroot()
    
    # Create a Pandas Excel writer
    writer = pd.ExcelWriter(xlsx_file, engine='xlsxwriter')
    
    # Iterate through each table in the XML file
    for table in root.findall('.//table'):
        # Convert each table to a DataFrame
        df = pd.DataFrame([{child.tag: child.text for child in row} for row in table])
        
        # Get the table name or use a default name
        table_name = table.get('name', f'Table_{root.findall(".//table").index(table) + 1}')
        
        # Write the DataFrame to a sheet in the Excel file
        df.to_excel(writer, sheet_name=table_name, index=False)
    
    # Save the Excel file
    writer.close()
```

```python
def xml_to_excel(xml_file_path, excel_file_path):
  """
  Extracts data from an XML file with multiple tables and writes them to separate sheets in an Excel file.

  Args:
    xml_file_path: Path to the XML file.
    excel_file_path: Path to the output Excel file.
  """
  try:
    # Parse the XML file
    tree = ET.parse(xml_file_path)
    root = tree.getroot()

    # Create an empty Excel writer
    writer = pd.ExcelWriter(excel_file_path, engine='openpyxl')

    # Iterate through each table in the XML (adjust XPath based on your XML structure)
    for table_tag in root.findall('.//table'): 
      table_data = []
      for row in table_tag.findall('tr'):
        row_data = []
        for cell in row.findall('td'):
          row_data.append(cell.text.strip() if cell.text else '') 
        table_data.append(row_data)

      # Create a DataFrame from the extracted data
      table_df = pd.DataFrame(table_data)

      # Get the table name (optional)
      table_name = table_tag.get('name') 
      if not table_name:
        table_name = f'Table_{table_data.index(table_data)+1}' 

      # Write the DataFrame to a new sheet in the Excel file
      table_df.to_excel(writer, sheet_name=table_name, index=False)

    writer.close()
    print(f"Data extracted from {xml_file_path} and written to {excel_file_path}")

  except Exception as e:
    print(f"An error occurred: {e}")
```

```python

# Example usage
xml_file_path = file  # Replace with the actual path to your XML file
excel_file_path = "output.xlsx"
xml_to_excel(xml_file_path, excel_file_path)
```
