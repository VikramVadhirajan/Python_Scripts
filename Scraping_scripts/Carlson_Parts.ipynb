```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from time import sleep
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.support import expected_conditions as EC
import pandas as pd
import datetime
import warnings
from tqdm import tqdm
import numpy as np
from IPython.display import clear_output
# Ignore all warnings (not recommended in general)
warnings.filterwarnings("ignore")
import os
```

```python
path= 'C://chromedriver.exe'
driver=webdriver.Chrome()
wait=WebDriverWait(driver, 5)
driver.get('https://carlsoncatalog.com/')
driver.maximize_window()
```

```python
cols =['PartNumber'] #,'Review_Mentions','Standards'
df1 = pd.DataFrame(columns=cols)
df2 = pd.DataFrame(columns=cols)
count1=0
count2=0
```

```python
df = pd.read_excel(r"C:\Users\vikram.vadhirajan\OneDrive - Trico\FBG\01_DRA\00_Projects\99_Automation_Projects\92_Our Products\Buyers guide ACES File preparation missing.xlsx" , sheet_name='Buyers guide')
```

```python
for p in tqdm(range(3858,len(df))):
    # try:
    #     wait.until(EC.presence_of_element_located(( By.XPATH, "//span[contains(@class, 'x-btn-inner') and text()='Close']").click()
    # except:
    #     pass
    searchbox=wait.until(EC.presence_of_element_located((By.CSS_SELECTOR, '[role="textbox"]')))
    searchbox.click()
    searchbox.clear()
    searchbox.send_keys(df["Part_Number"][p])
    print(p,":",df["Part_Number"][p]," | Rows:",count1)
    searchbox.send_keys(Keys.ENTER)
    sleep(3)
    errormessage=wait.until(EC.presence_of_element_located((By.CLASS_NAME, "x-window-text"))).text
    if "No part(s) found matching" not in errormessage:
        wait.until(EC.presence_of_element_located((By.CSS_SELECTOR, "a.PartNumberLink"))).click()
        sleep(4)
        wait.until(EC.presence_of_element_located((By.XPATH,"//span[contains(@class, 'x-tab-inner') and text()='Buyers Guide']/ancestor::span[contains(@class, 'x-tab-button')]"))).click()
        sleep(2)
        Pages = int(wait.until(EC.presence_of_element_located((By.XPATH, "//div[contains(@class, 'x-toolbar-text') and contains(text(), 'of')]"))).text.split(" ")[-1])
        for page in range(Pages):
            sleep(3)
            rows = wait.until(EC.presence_of_all_elements_located((By.CSS_SELECTOR, "tr.x-grid-row")))
            for row in rows:
                # Find all the <div> elements inside this row
                cells = row.find_elements(By.CSS_SELECTOR, "div.x-grid-cell-inner")
                if len(cells) >= 5:  # We expect 5 columns based on your HTML

                    df1.loc[count1,'PartNumber']=df["Part_Number"][p]
                    df1.loc[count1,'Make']=cells[0].text
                    df1.loc[count1,"Model"]=cells[1].text
                    df1.loc[count1,"Year"]=cells[2].text
                    df1.loc[count1,"Position"]=cells[3].text
                    df1.loc[count1,"Description"]=cells[4].text
                    count1+=1
            try:
                wait.until(EC.presence_of_element_located((By.XPATH,"//span[contains(@class, 'x-tbar-page-next')]/ancestor::span[contains(@class, 'x-btn-button')]"))).click()
            except:
                pass
        wait.until(EC.presence_of_element_located(( By.XPATH, "//span[contains(@class, 'x-btn-inner') and text()='Close']"))).click()
        clear_output(wait=True)
    else:
        wait.until(EC.element_to_be_clickable((By.XPATH, "//span[text()='OK']/ancestor::a"))).click()
```
