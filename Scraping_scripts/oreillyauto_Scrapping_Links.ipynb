# 4. Defining the Dataframe and Extracting the data into the Dataframe

```python
Cols=["Links","Name"]
df = pd.DataFrame(columns=Cols)
count=0
df
```

```python
  
Partnames=driver.find_elements(By.CLASS_NAME, 'product__name')
links=driver.find_elements(By.CLASS_NAME,"product__link")
partnumbers=driver.find_elements(By.CLASS_NAME, 'js-ga-product-line-number')
costs=driver.find_elements(By.CLASS_NAME, 'js-ga-product-price')

for partname,partnumber,cost,link in zip(Partnames,partnumbers,costs,links):
    df.loc[count,"Links"]=link.get_attribute('href')
    df.loc[count,"Name"]=partname.text
    df.loc[count,"PartNumber"]=partnumber.text
    df.loc[count,"Cost"]=cost.text
    #df.loc[count,'Rating']=Rating
    count=count+1
```

```python
for i in range(10):

  
    Partnames=driver.find_elements(By.CLASS_NAME, 'product__name')
    links=driver.find_elements(By.CLASS_NAME,"product__link")
    partnumbers=driver.find_elements(By.CLASS_NAME, 'js-ga-product-line-number')
    costs=driver.find_elements(By.CLASS_NAME, 'js-ga-product-price')

    for partname,partnumber,cost,link in zip(Partnames,partnumbers,costs,links):
        df.loc[count,"Links"]=link.get_attribute('href')
        df.loc[count,"Name"]=partname.text
        df.loc[count,"PartNumber"]=partnumber.text
        df.loc[count,"Cost"]=cost.text
        #df.loc[count,'Rating']=Rating
        count=count+1
    driver.find_element(By.TAG_NAME,'body').send_keys(Keys.END)
    sleep(1)
    driver.find_element(By.CSS_SELECTOR,'[aria-label="Next Page"]').click()

   
```
