```python
cols2 =['Sl.No']
df2= pd.DataFrame(columns=cols2)
count2=0
cols3 =['Sl.No','OENumbers']
df3= pd.DataFrame(columns=cols3)
count3=0
```

```python
for i in tqdm(range(1629,len(df))):
    print(f"Begins with the item# {i}, {df['PartNumber'][i]}")
    driver.get(df['Link'][i])
    sleep(1)
    lk=df['Link'][i]
    pname=driver.find_element(By.ID,'lblProductDesc').text
    pn=driver.find_element(By.ID,'lblProductName').text
    OEs=driver.find_elements(By.CSS_SELECTOR,'[scope="row"]')
    Olist=[]
    for oe in OEs:
        Olist.append(oe.text)
#----------------------------------------------------------------------------------------------------------------------------
    ps=driver.find_elements(By.CSS_SELECTOR,'[style="width:100%;"]')
    for p in ps:
        lines=len(p.text.split('\n'))
        for l in range(lines):
            Attr=p.text.split('\n')[l].split(": ")[0]
            Value=p.text.split('\n')[l].split(": ")[1]
            df3.loc[count3,'Sl.No']=count3
            df3.loc[count3,'Sl.No']=count3
            df3.loc[count3,'PartNumber']=pn
            df3.loc[count3,'PartName']=pname
            df3.at[count3,'OENumbers']=Olist
            df3.loc[count3,Attr]=Value
            df3.loc[count3,'ProductDescription']=driver.find_element(By.CSS_SELECTOR,'[id="productDesc"]').text
            df3.loc[count3,'Link']=lk
        count3=count3+1
#----------------------------------------------------------------------------------------------------------------------------
    frame_0 = driver.find_element(By.ID,'ifDetailApp')
    driver.switch_to.frame(frame_0)
    for i in range(100):
        try:        
            cs=driver.find_elements(By.CLASS_NAME,'detail-app-row')
            for c in cs:
                    df2.loc[count2,'Sl.No']=count2
                    df2.loc[count2,'PartNumber']=pn
                    df2.loc[count2,'PartName']=pname
                    df2.loc[count2,'Year']=int(c.find_element(By.CSS_SELECTOR,f'[style="width:5%;"]').text)
                    df2.loc[count2,'Make']=c.find_element(By.CSS_SELECTOR,f'[style="width:13%;"]').text
                    df2.loc[count2,'Model']=c.find_element(By.CSS_SELECTOR,f'[style="width:17%;"]').text
                    df2.loc[count2,'Positions']=c.find_elements(By.XPATH,"//td[@style='width:17%;']")[1].text
                    df2.loc[count2,'Configuration']=c.find_element(By.CSS_SELECTOR,f'[style="width:21%;"]').text
                    df2.loc[count2,'ApplicationNotes']=c.find_element(By.CSS_SELECTOR,f'[style="width:27%;"]').text            
                    count2=count2+1   
            driver.find_element(By.ID,"pagingBottom_nextButton").click()
            sleep(2)
        except:
            break
    clear_output(wait=True)
```
