```python
# Wait for the button to be clickable
button_locator = (By.CLASS_NAME, 'p-autocomplete-dropdown')
button = WebDriverWait(browser, 10).until(EC.element_to_be_clickable(button_locator))

# Click the button
button.click()
 

panel_locator = (By.CLASS_NAME, 'p-autocomplete-panel')
panel = WebDriverWait(browser, 10).until(EC.presence_of_element_located(panel_locator))

# Find all the items in the list
items_locator = (By.CSS_SELECTOR, '.p-autocomplete-items .ng-star-inserted')
items = panel.find_elements(*items_locator)

# Extract text from each item
item_texts = [item.text for item in items]
item_texts=set(item_texts)
item_texts=list(item_texts)
item_texts.sort()
# Print or use the item texts as needed
print(item_texts)
```

```python
cols=['PartNumber','Product_Name','Make','Models','Model_Years','Start_Year','End_Year']
df = pd.DataFrame(columns=cols)
df

```

```python
count=0
for j in range(len(item_texts)):
    inputkeys=item_texts[j]
    #print(inputkeys)
    input_element = browser.find_element(By.CLASS_NAME,'p-autocomplete-input')
    input_element.click()
    # Clear any existing text in the input field (optional, if needed)
    input_element.clear()
    # Type the target value into the input field
    input_element.send_keys(inputkeys)
    input_element.send_keys(Keys.ENTER)
    #print('entered')
    table_element = browser.find_element(By.XPATH, "//tbody[@_ngcontent-ng-c3462382488='']")
    # Find all rows in the table
    Model = table_element.find_elements(By.TAG_NAME, "th")
    Model_Year = table_element.find_elements(By.TAG_NAME, "td")
    for i in range(1,len(Model)):
            # This has to be defined. 
            df.loc[count,'Make']=item_texts[j]
            df.loc[count,'Product_Name']=Product_Name
            df.loc[count,'PartNumber']=ItemCode
            df.loc[count,'Models']=Model[i].text
            df.loc[count,'Model_Years']= Model_Year[i].text
            count=count+1            
    sleep(1)
    print('Manufacturers-',j+1, 'Models-',count)    
    input_element.clear()
    
    
```

```python
df
```

```python
df['Model_Years']=df['Model_Years'].str.split(',')
df=df.explode('Model_Years')
df = df.drop_duplicates()
df = df.reset_index(drop=True)
```

```python
for i in range(len(df)):
    if "-" in df['Model_Years'][i]:
        #print(i)
        year=df['Model_Years'].str.split('-')[i]
        startyear=year[0]
        endyear=year[1]
        df.loc[i,'Start_Year']=startyear
        df.loc[i,'End_Year']=endyear
    if "-" not in df['Model_Years'][i]:
        #print (i)
        df.loc[i,'Start_Year']=df['Model_Years'][i]
        df.loc[i,'End_Year']=df['Model_Years'][i]
```

```python
df['Start_Year']=df['Start_Year'].str.replace(" ","").astype(int)
df['End_Year']=df['End_Year'].str.replace(" ","").astype(int)
```

```python
del df['Model_Years']
df
```
