```python
cols =['PartNumber'] #,'Review_Mentions'
dfIC = pd.DataFrame(columns=cols)
countIC=0
df_a = pd.DataFrame(columns=cols)
counta=0
```

```python
for i in tqdm(range(len(df))):
    driver.get(df['Link'][i])
    print("Partnumber:",df['PartNumber'][i],"the value of i is",i)
    sleep(2)
    driver.find_elements(By.CLASS_NAME,"border-transparent")[1].click() # Application
    sleep(2)
    ICList=driver.find_element(By.CLASS_NAME,'gap-4').text.split("\n")
    for ic in range(len(ICList)):
        Cust=ICList[ic].rsplit(' ', 1)[0]
        ICNum=ICList[ic].rsplit(' ', 1)[1]
        dfIC.loc[countIC,'PartNumber']=df['PartNumber'][i] 
        dfIC.loc[countIC,"Customer"]=Cust
        dfIC.loc[countIC,"InterchangeNumber"]=ICNum
        countIC=countIC+1
    driver.find_elements(By.CLASS_NAME,"border-transparent")[1].click() # Application
    sleep(2)
    Years=driver.find_elements(By.CSS_SELECTOR,'[x-text="application.year"]')
    Makes=driver.find_elements(By.CSS_SELECTOR,'[x-text="application.make"]')
    Models=driver.find_elements(By.CSS_SELECTOR,'[x-text="application.model"]')
    Engines=driver.find_elements(By.CSS_SELECTOR,'''[x-text="application.displacement_liters + 'L'"]''')
    for year, make, model, engine in zip(Years,Makes, Models, Engines):
        df_a.loc[counta,'PartNumber']=df['PartNumber'][i]   
        df_a.loc[counta,'Year']=year.text
        df_a.loc[counta,'Make']=make.text
        df_a.loc[counta,'Model']=model.text
        df_a.loc[counta,'Engine']=engine.text
        counta=counta+1
    clear_output(wait=True)
```
