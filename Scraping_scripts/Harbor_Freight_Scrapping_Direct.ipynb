# 3. Setting Webdriver and Website Specific Information

```python
content=f.read()
List=content.split(Linebraker)
print(len(List))
```

# 4. Defining the Dataframe and Extracting the data into the Dataframe

```python
cols =['Sl.No'] #,'Review_Mentions'
df = pd.DataFrame(columns=cols)
```

```python
count=0

for i in range(1,len(List)):
    df.loc[count,'Sl.No']=i
    try:
        df.loc[count,'Current Price']=List[i].split('$</sup>')[1].split ('</sup></span>')[0].replace('<sup>',".").replace(',',"")
    except:
        df.loc[count,'Current Price']=0
    try:
        Rating=List[i].count('stars__active--abXZeh')+int(List[i].split('style="width: ')[1].split ('%;"></i>')[0])/100
        df.loc[count,'Ratings']=Rating
    except:
        df.loc[count,'Ratings']=0
    try:
        df.loc[count,'No of Ratings']=List[i].split('<span>(')[1].split (')</span>')[0].replace(',',"")
    except:
        df.loc[count,'No of Ratings']=0
    df.loc[count,'Name']=List[i].split('grid__name--IqXL92">')[1].split ('</p><div>')[0]
    try:
        link=List[i].split(' href="')[1].split ('"><div')[0]
    except:
        link="-"
    df.loc[count,'Link']=prefix+link
    df.loc[count,'Brand']=List[i].split('brand--LGWiQh">')[1].split ('</span>')[0]
    df.loc[count,'Source']=Source
    count=count+1
```

# 5. Post Processing Data and Exporting

```python
df['Sl.No']=df['Sl.No'].astype(int)
df['Current Price']=(df['Current Price']).astype(float)
df['Ratings']=(df['Ratings']).astype(float)
df['No of Ratings']=(df['No of Ratings']).astype(float)
```

```python
cols=["Sl.No","Name","Current Price","Ratings","No of Ratings",'Brand', 'Link',"Source"]
df = df[cols]
df.info()
```

```python
df.to_excel(OFolder+'\\'+f'{Source}ProductDetails_'+search_text+'.xlsx', index=False)
```
