```python
count=0
for i in range(len(df)):
    Alist=[]
    q=(len(df["Coverage"][i].split("\n")))
    for j in range(q):
        df1.loc[count,'Sl.No']=count
        df1.loc[count,"Make"]=(df["Coverage"][i].split("\n")[j].split(" ",1)[0])
        df1.loc[count,"Model"]=(df["Coverage"][i].split("\n")[j].split("(",1)[0].split(" ",1)[1])
        df1.at[count,"Model_Years"]=(df["Coverage"][i].split("\n")[j].split("(",1)[1].replace("(","").replace(")","").replace(" ","").split(","))
        df1.loc[count,"Part Number"]=df["Part Number"][i]
        df1.loc[count,"Link"]=df["Link"][i]
        count=count+1
```

```python
for i in range(len(df1)):
    if "-" in df1['Model_Years'][i]:
        #print(i)
        year=df1['Model_Years'].str.split('-')[i]
        startyear=year[1]
        endyear=year[0]
        df1.loc[i,'Start_Year']=startyear
        df1.loc[i,'End_Year']=endyear
    if "-" not in df1['Model_Years'][i]:
        #print (i)
        df1.loc[i,'Start_Year']=df1['Model_Years'][i]
        df1.loc[i,'End_Year']=df1['Model_Years'][i]
```

```python
Nprefix="20"
Oprefix="19"
for i in range(len(df1)):

    if int(df1['Start_Year'][i]) <24:
        df1.loc[i,'Start_Year']=int(Nprefix+df1['Start_Year'][i])
    else:
        df1.loc[i,'Start_Year']=int(Oprefix+df1['Start_Year'][i])
```

```python
Nprefix="20"
Oprefix="19"
for i in range(len(df1)):

    if int(df1['End_Year'][i]) <24:
        df1.loc[i,'End_Year']=int(Nprefix+df1['End_Year'][i])
    else:
        df1.loc[i,'End_Year']=int(Oprefix+df1['End_Year'][i])
```

```python
for i in range (len(df1)):    
    df1.at[i,"Years"]=list(range(df1["Start_Year"][i],(df1["End_Year"][i]+1)))
    
df1=df1.explode('Years')
```

```python
df1["Application"]=df1["Years"].astype(str)+" "+df1["Make"]+" "+df1["Model"]
df1
```

```python
cols=["Part Number","Make","Model","Years","Application","Link"]
df1=df1[cols]
```

```python
count=0
for i in range(len(df)):
    Alist=[]
    d=df['Spec'][i].replace(":\n",":").split("\n")
    for k in range(len(d)):
        if len(d[k])>2:
            Alist.append(d[k])
    df2.loc[count,"Part Number"]=df["Part Number"][i]
    df2.loc[count,"Link"]=df["Link"][i]
    df2.at[count,"Attributes"]=Alist
    count=count+1
```

```python
dfAtt=df2[['Part Number','Attributes',"Link"]]
dfAtt=dfAtt.explode('Attributes')
dfAtt=dfAtt.explode('Attributes')
dfAtt[['Attributes', 'Value']] = dfAtt['Attributes'].str.split(':',1, expand=True)
dfffff=dfAtt.pivot(index='Part Number',columns='Attributes')['Value']
dfffff
```

```python
cols=['Part Number','Attributes',"Value","Link"]
dfAtt=dfAtt[cols]
```

```python
summary_df = pd.pivot_table(dfAtt, values='Value', index='Attributes',aggfunc='count').reset_index()
summary_df =summary_df.sort_values(by='Value',ascending=False)
summary_df=summary_df.rename(columns ={'Value':'No Products contains this Attribute'})
summary_df
```
