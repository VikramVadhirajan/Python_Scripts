```python
for make in range(9,21):
    driver.get(home)
    sleep(2)
    List=wait.until(EC.presence_of_all_elements_located((By.CLASS_NAME, "font-semibold")))
    m=List[make].text
    print(m)
    List[make].click()
    sleep(2)
    search=wait.until(EC.presence_of_element_located((By.ID,"main_search_10")))
    search.click()
    search.send_keys(search_text)
    search.send_keys(Keys.ENTER)
    sleep(2)
    Varnum=1
    for i in range(16):
        sleep(10)
        Alles=wait.until(EC.presence_of_all_elements_located((By.CSS_SELECTOR,'[class="catalog-product marketplace-product row "]')))
        Names=wait.until(EC.presence_of_all_elements_located((By.CLASS_NAME,"product-title")))
        # PartNumbers=wait.until(EC.presence_of_all_elements_located((By.CLASS_NAME,"catalog-product-id")))
        # Links=wait.until(EC.presence_of_all_elements_located((By.CLASS_NAME,"product-title")))
        # Replaces=wait.until(EC.presence_of_all_elements_located((By.CSS_SELECTOR,'[class="catalog-product-more-in-category catalog-info-row"]')))
        # Fits=wait.until(EC.presence_of_all_elements_located((By.CLASS_NAME,"fitment-makes ")))
        sleep(5)
        # for All, Name, partnumber, replaces, fit in zip (Alles, Names, PartNumbers, Replaces, Fits):
        for All, Name in zip (Alles, Names):
            df.loc[count,'Make']=m
            df.loc[count,'#Variants']=Varnum
            df.loc[count,"Sl.No."]=count
            df.loc[count,"PartName"]=All.text.split("\n")[0]
            try:
                df.loc[count,"PartNumber"]=All.text.split("Part Number: ")[1].split("\n")[0]
            except:
                df.loc[count,"PartNumber"]=All.text.split("PART NUMBER: ")[1].split("\n")[0]
            df.loc[count,"Link"]=Name.find_element(By.CSS_SELECTOR,"a").get_attribute('href').split("?")[0]
            try:
                df.loc[count,"Replaces"]=All.text.split("Replaces: ")[1].split("\n")[0]
            except:
                df.loc[count,"Replaces"]=""
            try:
                df.loc[count,"Fitment"]=All.text.split("Fits: ")[1].split("\n")[0]
            except:
                df.loc[count,"Fitment"]=""
            try:
                df.loc[count,"Position"]=All.text.split("Positions: ")[1].split("\n")[0]
            except:
                df.loc[count,"Position"]=""
            
            count=count+1
            Varnum=Varnum+1
        driver.find_element(By.CSS_SELECTOR,f'[data-page="{i+2}"]').click()
        sleep(2)
    driver.get(home)
```
