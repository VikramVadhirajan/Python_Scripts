```python
import datetime
from datetime import timedelta

import numpy as np
from IPython.display import clear_output

timestamp=[]
timediff=[]
def timeremaining(balanceitem,i):
    timestamp.insert(i,datetime.datetime.now())
    if i>=1:
        diff=timestamp[i]-timestamp[i-1]
        timediff.insert(i,diff.total_seconds())		
        remainingtime=np.median(timediff)*balanceitem
        millis=int(remainingtime)		
        #print(millis)
        seconds=(millis)%60
        seconds = int(seconds)
        minutes=(millis/(60))%60
        minutes = int((minutes)) #math.floor
        hours=(millis/(60*60))%24
        hours=int(hours)
        print(f'Item looped: {i+1} \nItems Remining: {balanceitem-1} and \nTime Remaining: {hours}:{minutes}:{seconds}')
        etc=datetime.datetime.now()+timedelta(hours=hours,minutes=minutes,seconds=seconds)
        print(f'Estimated time of completion:{etc}')        
        clear_output(wait=True)
```

```python
for s in tqdm(range(8,totpage)):
    print(s, count)
    partcount=len(wait.until(EC.presence_of_all_elements_located((By.CLASS_NAME,"ag-row-position-absolute"))))
    for i in range(partcount):
        df_pa.loc[count,'SupplierPartNumber']=(driver
                                               .find_elements(By.CLASS_NAME,"ag-row-position-absolute")[i]
                                               .find_element(By.CSS_SELECTOR,'[col-id="articleNo"]')
                                               .text
                                               )
        df_pa.loc[count,'Supplier']=(driver
                                     .find_elements(By.CLASS_NAME,"ag-row-position-absolute")[i]
                                     .find_element(By.CSS_SELECTOR,'[col-id="description"]')
                                     .find_elements(By.CLASS_NAME,"font-weight-bold")[0]
                                     .text
                                    )
        df_pa.loc[count,'ProductName']=(driver
                                        .find_elements(By.CLASS_NAME,"ag-row-position-absolute")[i]
                                        .find_element(By.CSS_SELECTOR,'[col-id="description"]')
                                        .find_elements(By.CLASS_NAME,"font-weight-bold")[1]
                                        .text
                                        )
        Atts=(driver
              .find_elements(By.CLASS_NAME,"ag-row-position-absolute")[i]
              .find_element(By.CSS_SELECTOR,'[col-id="description"]')
              .find_elements(By.CLASS_NAME,'article-criteria')
              )
        Attributes=[]
        for attri in Atts:
            CN=attri.text.split(":")[0]
            CV=attri.text.split(":")[1]       
            df_pa.loc[count,CN]=CV
        df_pa.at[count,'Attributes']=Attributes
        df_pa.loc[count,'Status']=(driver
                                   .find_elements(By.CLASS_NAME,"ag-row-position-absolute")[i]
                                   .find_element(By.CSS_SELECTOR,'[col-id="stateText"]')
                                   .text
                                   )
        df_pa.loc[count,'Link']=prefix+(driver
                                        .find_elements(By.CLASS_NAME,"ag-row-position-absolute")[i]
                                        .get_attribute('row-id')
                                        .replace(']-[','/')
                                        .replace("[","")
                                        .replace("]","")
                                        )+suffix
        count=count+1
    clear_output(wait=True)
    driver.find_element(By.CSS_SELECTOR,'[aria-label="Next Page"]').click()
    sleep(10)
```
