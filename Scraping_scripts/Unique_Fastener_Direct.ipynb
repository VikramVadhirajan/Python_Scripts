# 4. Defining the Dataframe and Extracting the data into the Dataframe

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from time import sleep
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.support import expected_conditions as EC
import pandas as pd
import datetime
import warnings
from tqdm import tqdm
import numpy as np
from IPython.display import clear_output
# Ignore all warnings (not recommended in general)
warnings.filterwarnings("ignore")
import os
```

```python
path= 'C://chromedriver.exe'
driver=webdriver.Chrome()
wait=WebDriverWait(driver, 5)
```

```python
driver.get('https://uniquefasteners.com/our-catalog/')
```

```python
cols =['Sl.No'] #,'Review_Mentions'
df = pd.DataFrame(columns=cols)
count=0
```

```python
cols =['Sl.No'] #,'Review_Mentions'
df1 = pd.DataFrame(columns=cols)
count1=0
```

```python
prod_cat1=driver.find_elements(By.CLASS_NAME, "int-card-body")
links=driver.find_elements(By.CSS_SELECTOR, 'div.int-card-body a.int-button')
for i, j in zip(prod_cat1,links):
    df.loc[count,'Product']=i.text.split("\n")[0]
    df.loc[count,'Link']=j.get_attribute('href')
    count= count+1
```

```python
df
```

```python
driver.find_element(By.CLASS_NAME,"elementor-size-sm").get_attribute('href')
```

```python
for i in range(len(df['Link'])):
    driver.get(df['Link'][i])
    sleep(3)
    Products=driver.find_elements(By.CLASS_NAME,"elementor-size-default")
    Links=driver.find_elements(By.CLASS_NAME,"elementor-size-sm")
    for p,q in zip(Products,Links):
        df1.loc[count1,"Product_Category"]=df['Product'][i]
        df1.loc[count1,'Product']=p.text
        df1.loc[count1,'Link']=q.get_attribute('href')
        count1=count1+1
```
