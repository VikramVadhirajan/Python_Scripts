```python
cols=['Part No.']
df = pd.DataFrame(columns=cols)
df
count=0

linkprefix="https://web.tecalliance.net/opticat/qa/parts/"
linksuffix="/ap-detail"
```

```python
from tqdm import tqdm
for i in tqdm(range(len(List))):#len(List))):
    if "|" not in List[i]:
        searchbutton=driver.find_element(By.ID,"part-search-input")
        searchbutton.click()
        searchbutton.send_keys(List[i])
        driver.find_element(By.CSS_SELECTOR,'[ta-name="search-button"]').click()
        sleep(2)
        try:
            Nodata=driver.find_element(By.CSS_SELECTOR, '[class="text-nowrap ng-star-inserted"]').text 
            df.loc[count,"SearchText"]=List[i]
            df.loc[count,"Competitor"]=Nodata
            df.loc[count,"Link"]=driver.current_url
            count=count+1
        except:
            statuses=wait.until(EC.presence_of_all_elements_located((By.CLASS_NAME, "article-status-cell")))
            Contents=driver.find_elements(By.CLASS_NAME, "generic-article") 
            PNS=driver.find_elements(By.CLASS_NAME, "text-primary")
            CNS=driver.find_elements(By.CLASS_NAME, "font-weight-bold")
            searchtext=driver.find_element(By.CLASS_NAME, "ng-star-inserted") 
            Productcodes=driver.find_elements(By.CSS_SELECTOR,'div[role="row"]')
            for status, content, pn,cn, productcode in zip(statuses,Contents,PNS,CNS,Productcodes[1:]):
                df.loc[count,"SearchText"]=List[i]
                df.loc[count,"Competitor"]=cn.text
                df.loc[count,"Part No."]=pn.text
                #df.loc[count,"Status"]=status.text
                df.loc[count,"Name"]=content.text
                pc=productcode.get_attribute("row-id").replace("]-[","/").replace("[","").replace("]","")
                link=linkprefix+pc+linksuffix
                df.loc[count,"Link"]=driver.current_url
                df.loc[count,"ProductLink"]=link
                count=count+1
    else:
            df.loc[count,"SearchText"]=List[i]
            df.loc[count,"Competitor"]="To be Dealt separately"
            count=count+1
        
    clear_output(wait=True)
```
