# 4. Defining the Dataframe and Extracting the data into the Dataframe

```python
cols =['Links'] #,'Review_Mentions'
df1 = pd.DataFrame(columns=cols)
count1=0
```

```python
Details=driver.find_elements(By.CSS_SELECTOR,'[class="s-item__info clearfix"]')
for i in range(len(Details)):
    if i>=1:
        df1.loc[count1,"Name"]=Details[i].text.split("\n")[0]
        try:
            df1.loc[count1,"Current Price"]=float(Details[i].text.split("$")[1].split("\n")[0])
        except:
            df1.loc[count1,"Current Price"]=Details[i].text.split("$")[1].split("\n")[0]
        try:
            df1.loc[count1,"List Price"]=float(Details[i].text.split("Was: $")[1].split("  ")[0])
        except:
            df1.loc[count1,"List Price"]=0
        try:
            df1.loc[count1,"Sales_LastMonth"]=int(Details[i].text.split("sold")[0].split("\n")[-1])
        except:
            df1.loc[count1,"Sales_LastMonth"]=0
        df1.loc[count1,"Links"]=Details[i].find_element(By.CLASS_NAME,"s-item__link").get_attribute("href")
        df1.loc[count1,"ItemNumber"]=Details[i].text.split("Item: ")[1].split("\n")[0]
        try:
            df1.loc[count1,"Rating"]=Details[i].text.split(" out of 5")[0].split("\n")[-1]
        except:
            pass
        try:
            df1.loc[count1,"No of Ratings"]=Details[i].text.split("product rating")[0].split("\n")[-1]
        except:
            pass
        count1=count1+1
```

```python
for i in tqdm(range(length)):
    driver.get(df1['Links'][i])
    df.loc[count,'Links']=df1['Links'][i]
    sleep(1)

    delements=driver.find_elements(By.CLASS_NAME,"ux-labels-values__labels")
    velements=driver.find_elements(By.CLASS_NAME,"ux-labels-values__values")
    Alist=[]
    for d, v in zip(delements,velements):
        c=d.text+v.text
        s=d.text+":" +v.text
        if ":" in c:
            pass
        else:
            if s.startswith(':'):
                pass
            else:
                Alist.append(s)
    df['Attributes'] = df['Attributes'].astype('object')
    df.at[count,'Attributes']=Alist
    df.loc[count,"Source"]=Source
    df.loc[count,"Product"]=search_text
    count=count+1
    clear_output(wait=True)
```

# 5. Post Processing Data and Exporting

```python
df['Brand']=df["Attributes"].apply(lambda x: str(x).replace('[','').replace(']','').replace('\'', '').replace(' ','')).str.split('Brand:',expand=True)[1].str.split(',',expand=True)[0]
df['Material']=df["Attributes"].apply(lambda x: str(x).replace('[','').replace(']','').replace('\'', '').replace(' ','')).str.split('Material:',expand=True)[1].str.split(',',expand=True)[0]
df['Part Number']=df["Attributes"].apply(lambda x: str(x).replace('[','').replace(']','').replace('\'', '').replace(' ','')).str.split("ManufacturerPartNumber:",expand=True)[1].str.split(',',expand=True)[0]
df['UPC Code']=df["Attributes"].apply(lambda x: str(x).replace('[','').replace(']','').replace('\'', '').replace(' ','')).str.split('UPC:',expand=True)[1].str.split(',',expand=True)[0]
```

```python
dfAtt=Mergeddf[['Links','Attributes']]
dfAtt=dfAtt.explode('Attributes')
dfAtt[['Attributes', 'Value']] = dfAtt['Attributes'].str.split(':',1, expand=True)
```

```python
summary_df = pd.pivot_table(dfAtt, values='Value', index='Attributes',aggfunc='count').reset_index()
summary_df =summary_df.sort_values(by='Value',ascending=False)
summary_df=summary_df.rename(columns ={'Value':'No Products contains this Attribute'})
summary_df
```
